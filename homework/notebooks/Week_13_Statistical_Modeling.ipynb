# Week 13: Statistical Modeling

Understanding the statistical underpinnings of data models, from hypothesis testing to regression analysis using `statsmodels`.

## 1. Hypothesis Testing with `scipy.stats`

Using p-values and confidence intervals to validate data observations.

```python
from scipy import stats
import numpy as np

data1 = np.random.normal(0, 1, 100)
data2 = np.random.normal(0.5, 1, 100)

# Perform t-test
t_stat, p_val = stats.ttest_ind(data1, data2)
print(f"p-value: {p_val}")
```

## 2. Regression with `statsmodels`

Providing more detailed statistical insights than `scikit-learn`.

```python
import statsmodels.api as sm

X = sm.add_constant(np.array([1, 2, 3]))
y = np.array([2, 4, 6])
model = sm.OLS(y, X).fit()
print(model.summary())
```
