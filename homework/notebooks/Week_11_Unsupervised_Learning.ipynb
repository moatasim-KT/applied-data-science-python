# Week 11: Unsupervised Learning

Exploring techniques to find patterns in unlabeled data.

## 1. Clustering

Grouping similar data points using K-Means.

```python
from sklearn.cluster import KMeans
import numpy as np

X = np.array([[1, 2], [1, 4], [1, 0], [10, 2], [10, 4], [10, 0]])
kmeans = KMeans(n_clusters=2).fit(X)
print(kmeans.labels_)
```

## 2. Dimensionality Reduction

Reducing the number of features while preserving the essential structure of the data using PCA.

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=1)
X_reduced = pca.fit_transform(X)
print(f"Reduced shape: {X_reduced.shape}")
```
