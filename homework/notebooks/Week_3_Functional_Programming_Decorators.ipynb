# Week 3: Functional Programming & Decorators

This week, we explore functional programming paradigms in Python, which help in writing clean, modular code for data manipulation pipelines.

## 1. Lambda, Map, and Filter

Functional constructs for data transformation.

```python
data = [1, 2, 3, 4, 5]

# Lambda and map
squared = list(map(lambda x: x**2, data))
print(f"Squared: {squared}")

# Lambda and filter
evens = list(filter(lambda x: x % 2 == 0, data))
print(f"Evens: {evens}")
```

## 2. Decorators

Decorators allow us to wrap another function to extend its behavior without permanently modifying it.

```python
import functools
import time

def timer(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end - start:.4f}s")
        return result
    return wrapper

@timer
def process_data(n):
    return sum(range(n))

process_data(1000000)
```
