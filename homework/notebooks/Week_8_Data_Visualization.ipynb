# Week 8: Data Visualization

Visualization is crucial for understanding data. We will cover Matplotlib for control and Seaborn for statistical plotting.

## 1. Matplotlib Basics

Matplotlib's object-oriented interface provides fine-grained control.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Sine Wave")
plt.show()
```

## 2. Statistical Visualization with Seaborn

Seaborn makes complex statistical plots easy to generate.

```python
import seaborn as sns

data = sns.load_dataset("tips")
sns.scatterplot(data=data, x="total_bill", y="tip", hue="sex")
plt.show()
```
