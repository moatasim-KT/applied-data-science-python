## 5. Tuples and Unpacking

Tuples are immutable sequences. They are often used for fixed-size collections of heterogeneous data.

```python
# Unpacking
point = (10, 20)
x, y = point  # Unpacking tuple
print(f"X: {x}, Y: {y}")

# Extended unpacking
first, *rest = [1, 2, 3, 4, 5]
print(f"First: {first}, Rest: {rest}")
```

## 6. Generators and Lazy Evaluation

Generators allow you to iterate over large data sets without loading everything into memory at once.

```python
# Generator expression
gen = (x**2 for x in range(1000000))
print(next(gen))
print(next(gen))
```

## 7. Itertools

The `itertools` library provides efficient tools for constructing complex iterators.

```python
import itertools

# Product of iterables
for p in itertools.product([1, 2], ['a', 'b']):
    print(p)
```
