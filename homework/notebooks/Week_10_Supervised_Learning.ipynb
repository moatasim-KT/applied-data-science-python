# Week 10: Supervised Learning - Algorithms

A deep dive into supervised learning algorithms: from linear models to ensemble methods.

## 1. Linear & Logistic Regression

Foundational models for regression and classification tasks.

```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1], [2], [3]])
y = np.array([2, 4, 6])

model = LinearRegression()
model.fit(X, y)
print(f"Slope: {model.coef_[0]}, Intercept: {model.intercept_}")
```

## 2. Tree-Based Models

Decision Trees and Random Forests handle non-linear relationships well.

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(n_estimators=100)
rf.fit(X, y)
print(f"Prediction: {rf.predict([[4]])}")
```
