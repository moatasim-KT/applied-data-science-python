# Week 7: Pandas Part II - Advanced Reshaping

In this lesson, we cover advanced data manipulation techniques, including merging, grouping, and reshaping data.

## 1. Merging and Joining

Combining DataFrames is a fundamental operation, analogous to SQL joins.

```python
import pandas as pd

df1 = pd.DataFrame({'Key': ['A', 'B'], 'Value1': [1, 2]})
df2 = pd.DataFrame({'Key': ['A', 'B'], 'Value2': [3, 4]})

# Merge on key
merged = pd.merge(df1, df2, on='Key')
print(merged)
```

## 2. GroupBy and Aggregation

The Split-Apply-Combine pattern is powerful for summarizing large datasets.

```python
df = pd.DataFrame({'Category': ['A', 'A', 'B', 'B'], 'Value': [1, 2, 3, 4]})

# Group by Category and sum
grouped = df.groupby('Category').sum()
print(grouped)
```

## 3. Reshaping Data

Transforming data between wide and long formats using `pivot` and `melt`.

```python
df = pd.DataFrame({'Date': ['2026-05-01', '2026-05-01'], 'City': ['NY', 'LA'], 'Temp': [70, 80]})

# Pivot table
pivoted = df.pivot(index='Date', columns='City', values='Temp')
print(pivoted)
```
