# Week 4: Concurrency & Asynchronous I/O for Data Ingestion

In data science, we often deal with slow I/O operations (fetching data from APIs, reading large files). This week, we learn to parallelize these tasks.

## 1. Multi-threading vs. Multi-processing

Due to Python's Global Interpreter Lock (GIL), multi-threading is effective for I/O-bound tasks, while multi-processing is required for CPU-bound tasks.

```python
import concurrent.futures
import time

def slow_io_task(n):
    time.sleep(1)
    return n * n

# ThreadPoolExecutor for I/O bound
with concurrent.futures.ThreadPoolExecutor() as executor:
    results = list(executor.map(slow_io_task, range(5)))
    print(f"Results (threaded): {results}")
```

## 2. Asynchronous Programming with `asyncio`

`asyncio` is a library to write concurrent code using the async/await syntax.

```python
import asyncio

async def async_io_task(n):
    await asyncio.sleep(1)
    return n * n

async def main():
    tasks = [async_io_task(n) for n in range(5)]
    results = await asyncio.gather(*tasks)
    print(f"Results (async): {results}")

# asyncio.run(main()) # Uncomment to run in a real environment
```
