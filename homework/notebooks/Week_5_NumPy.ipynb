# Week 5: NumPy: Memory-Efficient Computing

NumPy is the backbone of numerical computing in Python. Its efficiency comes from pre-compiled C routines and dense, homogenous memory layouts.

## 1. ndarray Memory Layout

Understanding memory layout (Row-major vs Column-major) is key for cache efficiency, a concept familiar from C++.

```python
import numpy as np

# Row-major (C-style, default)
a = np.array([[1, 2, 3], [4, 5, 6]], order='C')
print(f"C-order (strides): {a.strides}")

# Column-major (F-style)
b = np.array([[1, 2, 3], [4, 5, 6]], order='F')
print(f"F-order (strides): {b.strides}")
```

## 2. Advanced Indexing

Beyond basic slicing, NumPy supports advanced techniques for data extraction.

```python
arr = np.array([10, 20, 30, 40, 50])

# Boolean Masking
mask = arr > 30
print(f"Masked: {arr[mask]}")

# Fancy Indexing
indices = [0, 2, 4]
print(f"Fancy: {arr[indices]}")
```

## 3. Broadcasting

Broadcasting allows NumPy to perform operations on arrays of different shapes without explicit replication of data.

```python
A = np.array([[1, 2], [3, 4]])
b = np.array([10, 20])

# A (2x2) + b (1x2) -> Broadcating b to A's shape
print(A + b)
```
