# Week 6: Pandas Part I - Data Cleaning & Wrangling

Pandas is the primary tool for data manipulation in Python. We will focus on data ingestion, inspection, and cleaning.

## 1. Introduction to Series and DataFrames

DataFrames are the workhorse of Pandas, similar to tables in a SQL database.

```python
import pandas as pd
import numpy as np

# Creating a DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35], 'Salary': [50000, 60000, 70000]}
df = pd.DataFrame(data)
print(df.head())
```

## 2. Handling Missing Data

Real-world data is messy. Pandas provides robust tools to identify and deal with `NaN` (Not a Number) values.

```python
df_missing = df.copy()
df_missing.loc[0, 'Salary'] = np.nan

# Identify missing data
print(df_missing.isnull().sum())

# Fill missing data or drop rows
df_filled = df_missing.fillna(df_missing['Salary'].mean())
print(df_filled)
```

## 3. Filtering and Selection

Efficiently selecting subsets of data using label-based (`loc`) or positional (`iloc`) indexing.

```python
# Conditional filtering
subset = df[df['Age'] > 28]
print(subset)
```
